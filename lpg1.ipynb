## Step 1: Install and Import Required Libraries

```python
# Install NLTK if needed
# pip install nltk

import nltk
from nltk.tokenize import word_tokenize, WhitespaceTokenizer, WordPunctTokenizer, TreebankWordTokenizer
from nltk.tokenize import TweetTokenizer, MWETokenizer
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer
from nltk.corpus import wordnet
import warnings
warnings.filterwarnings('ignore')

# Download required data
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

## Step 2: Sample Text

```python
# Sample text for tokenization
text = "Hello, world! How are you? I'm learning NLP with NLTK. Let's do it!"
tweet = "Hey @John! #NLP is awesome. Can't wait for the weekend! :) www.example.com"

print("Original Text:", text)
print("\nTweet:", tweet)
```

## Step 3: Different Tokenization Methods

```python
# 1. WHITESPACE TOKENIZATION
print("1. WHITESPACE TOKENIZATION")
print("="*50)
whitespace_tokenizer = WhitespaceTokenizer()
whitespace_tokens = whitespace_tokenizer.tokenize(text)
print("Tokens:", whitespace_tokens)
print()
```

```python
# 2. PUNCTUATION-BASED TOKENIZATION (WordPunctTokenizer)
print("2. PUNCTUATION-BASED TOKENIZATION")
print("="*50)
punct_tokenizer = WordPunctTokenizer()
punct_tokens = punct_tokenizer.tokenize(text)
print("Tokens:", punct_tokens)
print()
```

```python
# 3. TREEBANK TOKENIZATION
print("3. TREEBANK TOKENIZATION")
print("="*50)
treebank_tokenizer = TreebankWordTokenizer()
treebank_tokens = treebank_tokenizer.tokenize(text)
print("Tokens:", treebank_tokens)
print()
```

```python
# 4. TWEET TOKENIZATION
print("4. TWEET TOKENIZATION")
print("="*50)
tweet_tokenizer = TweetTokenizer()
tweet_tokens = tweet_tokenizer.tokenize(tweet)
print("Tokens:", tweet_tokens)
print()
```

```python
# 5. MULTI-WORD EXPRESSION (MWE) TOKENIZATION
print("5. MULTI-WORD EXPRESSION (MWE) TOKENIZATION")
print("="*50)

mwe_tokenizer = MWETokenizer([('New', 'York'), ('San', 'Francisco')])
text_with_mwe = "I visited New York and San Francisco last year."
mwe_tokens = mwe_tokenizer.tokenize(word_tokenize(text_with_mwe))

print("Text:", text_with_mwe)
print("Tokens:", mwe_tokens)
print()
```

## Step 4: Stemming Techniques

```python
# Sample words for stemming
words = ['running', 'runs', 'runner', 'quickly', 'jumping', 'jumped', 'jumps', 'caring', 'cares']

print("STEMMING COMPARISON")
print("="*70)
print(f"{'Original':<15} {'Porter':<15} {'Snowball'}")
print("-"*70)
```

```python
# 1. PORTER STEMMER
porter_stemmer = PorterStemmer()

# 2. SNOWBALL STEMMER (supports multiple languages)
snowball_stemmer = SnowballStemmer('english')

# Compare both stemmers
for word in words:
    porter_stem = porter_stemmer.stem(word)
    snowball_stem = snowball_stemmer.stem(word)
    print(f"{word:<15} {porter_stem:<15} {snowball_stem}")
```

## Step 5: Lemmatization

```python
# LEMMATIZATION using WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

print("\nLEMMATIZATION")
print("="*70)
print(f"{'Original':<15} {'Lemmatized (Noun)':<20} {'Lemmatized (Verb)'}")
print("-"*70)

for word in words:
    noun_lemma = lemmatizer.lemmatize(word, pos=wordnet.NOUN)
    verb_lemma = lemmatizer.lemmatize(word, pos=wordnet.VERB)
    print(f"{word:<15} {noun_lemma:<20} {verb_lemma}")
```

## Step 6: Complete Pipeline Example

```python
# Complete text processing pipeline
print("\n\nCOMPLETE PIPELINE EXAMPLE")
print("="*70)

sample_text = "The runners are running quickly. They jumped over the fence."
print("\nOriginal Text:", sample_text)

# Step 1: Tokenize
tokens = word_tokenize(sample_text.lower())
print("\nTokenized:", tokens)

# Step 2: Stem
stemmed = [porter_stemmer.stem(word) for word in tokens]
print("\nAfter Stemming:", stemmed)

# Step 3: Lemmatize
lemmatized = [lemmatizer.lemmatize(word, pos=wordnet.VERB) for word in tokens]
print("\nAfter Lemmatization:", lemmatized)
```
